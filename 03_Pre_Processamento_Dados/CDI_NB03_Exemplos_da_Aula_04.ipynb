## 1. Agregação (Redução de Dados)
Combinar múltiplos objetos em um único.

```python
# Exemplo: vendas diárias durante 30 dias
np.random.seed(0)
vendas_diarias = np.random.randint(100, 500, 30)

# agregando para vendas mensais
vendas_mensais = np.sum(vendas_diarias)

print("Vendas diárias:", vendas_diarias)
print("Venda mensal (agregação):", vendas_mensais)
```

## 2. Amostragem
Selecionar subconjunto representativo de dados.

```python
# dataset grande
dados = np.arange(1000)

# amostragem aleatória simples
amostra = np.random.choice(dados, size=100, replace=False)

print("Tamanho original:", len(dados))
print("Tamanho da amostra:", len(amostra))
```

```python
# Amostragem estratificada simples
# simulando classes
classes = np.array([0]*500 + [1]*500)
np.random.shuffle(classes)
idx_classe0 = np.where(classes == 0)[0]
idx_classe1 = np.where(classes == 1)[0]

amostra0 = np.random.choice(idx_classe0, 50, replace=False)
amostra1 = np.random.choice(idx_classe1, 50, replace=False)

amostra_estratificada = np.concatenate([amostra0, amostra1])

print("ìndices da amostra estratificada:", amostra_estratificada)
```

## 3. Redução de Dimensionalidade (SVD / PCA)

```python
# matriz de dados
X = np.random.rand(20, 5)

# centralizando os dados
X_c = X - np.mean(X, axis=0)

# SVD
U, S, VT = np.linalg.svd(X_c)

# projeção para 2 dimensões
X_reduzido = np.dot(X_c, VT[:2].T)

print("Dimensão original:", X)
print("Dimensão reduzida:", X_reduzido)
```

## 4. Seleção de Subconjunto de Características
Selecionar atributos relevantes.

```python
# dataset
X = np.random.rand(100, 5)

# variância de cada atributo
variancias = np.var(X, axis=0)

# selecionar atributos com maior variância
top_features = np.argsort(variancias)[-3:]

X_selecionado = X[:, top_features]

print("Variâncias:", variancias)
print("Features selecionadas:", top_features)
print("Novo shape:", X_selecionado.shape)
```

## 5. Criação de Características

```python
# exemplo: densidade = massa / volume

massa = np.array([10, 20, 30])
volume = np.array([2, 4, 5])

densidade = massa / volume

print("Densidade:", densidade)
```

```python
# exemplo usando FFT (mapear para espaço de frequência)

sinal = np.sin(np.linspace(0, 10, 100))

fft = np.fft.fft(sinal)

print("Primeiros valores da FFT:", fft[:5])
```

## 6. Ponderação de Características

```python
X = np.random.rand(5,3)

pesos = np.array([0.2, 0.5, 1.5])

X_ponderado = X * pesos

print("Original:\n", X)
print("Ponderado:\n", X_ponderado)
```

## 7. Discretização

```python
dados = np.random.rand(20) * 100

# largura igual
bins = np.linspace(min(dados), max(dados), 4)

discretizado = np.digitize(dados, bins)

print("Dados:", dados)
print("Bins:", bins)
print("Discretizado:", discretizado)
```

## 8. Binarização (One-Hot simples)

```python
categorias = np.array([0,2,2,1,0,1,1,0])

num_classes = np.max(categorias) + 1

binarizado = np.eye(num_classes)[categorias]

print("Categorias:", categorias)
print("Binarizado:\n", binarizado)
```

## 9. Transformações de Variáveis

```python
x = np.array([1,2,256,2048,512])

log_x = np.log2(x)
sqrt_x = np.sqrt(x)
inv_x = 1/x

print("Original:", x)
print("log:", log_x)
print("sqrt:", sqrt_x)
print("1/x:", inv_x)
```

## 10. Padronização / Normalização

```python
dados = np.random.rand(100) * 100

media = np.mean(dados)
desvio = np.std(dados)

padronizado = (dados - media) / desvio

print("Média após padronização:", np.mean(padronizado))
print("Desvio padrão:", np.std(padronizado))
print(np.sum((dados - np.median(dados))/dados.size))
```
